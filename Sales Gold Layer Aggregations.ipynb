```python
%sql
-- Monthly sales trend
CREATE OR REPLACE TABLE workspace.sales_project.gold_monthly_sales AS
SELECT
  order_year,
  order_month,
  make_date(order_year, order_month, 1) AS month_start,
  ROUND(SUM(sales), 2)  AS total_sales,
  COUNT(DISTINCT order_id) AS total_orders
FROM workspace.sales_project.silver_sales
GROUP BY order_year, order_month;

-- Sales by region and category
CREATE OR REPLACE TABLE workspace.sales_project.gold_region_category AS
SELECT
  region,
  category,
  ROUND(SUM(sales), 2)  AS total_sales
FROM workspace.sales_project.silver_sales
GROUP BY region, category;

-- Top 10 products
CREATE OR REPLACE TABLE workspace.sales_project.gold_top_products AS
SELECT product_name, ROUND(SUM(sales), 2) AS total_sales
FROM workspace.sales_project.silver_sales
GROUP BY product_name
ORDER BY total_sales DESC
LIMIT 10;
```
